## without column transformer
Manual approach to transform the data. need to concat all the transformed columns together at the end

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df.isnull().sum()
```

```python
# adding simple imputer to ferver col
si = SimpleImputer()
X_train_fever = si.fit_transform(X_train[["fever"]])

X_test_fever = si.transform(X_test[["fever"]])
```

```python
X_train_fever.shape
```

```python
oe = OrdinalEncoder(categories=[["Mild","Strong"]])
X_train_cough = oe.fit_transform(X_train[["cough"]])
X_test_cough = oe.fit_transform(X_test[["cough"]])

X_train_cough.shape
```

```python
X_test_cough
```

```python
ohe=OneHotEncoder(drop="first",sparse_output=False)
X_train_gender_city=ohe.fit_transform(X_train[["gender","city"]])
X_test_gender_city=ohe.transform(X_test[["gender","city"]])

X_train_gender_city.shape

```

```python
X_train_gender_city
```

```python
# so we trasformed all except the age column, we can concatenate all the transformed columns together
# but first we need to seperate age column from X_train and X_test
X_train_age=X_train.drop(columns=["gender","city","fever","cough"]).values
X_test_age=X_test.drop(columns=["gender","city","fever","cough"]).values

X_train_age.shape
```

```python
X_train_age
```

```python
X_train_transformed=np.concatenate([X_train_age,X_train_gender_city, X_train_fever,X_train_cough],axis=1)
X_test_transformed=np.concatenate([X_test_age,X_test_gender_city, X_test_fever,X_test_cough],axis=1)

X_train_transformed.shape
```

## With column transformer

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(
    transformers=[
        ("tnf1", SimpleImputer(), ["fever"]),
        ("tnf2", OrdinalEncoder(categories=[["Mild","Strong"]]), ["cough"]),
        ("tnf3", OneHotEncoder(drop="first",sparse_output=False), ["gender","city"])
    ],
    remainder="passthrough"
)
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.fit_transform(X_train)
```
