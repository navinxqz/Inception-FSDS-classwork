```python
import pandas as pd
import seaborn as sns
```

```python
titanic = pd.read_csv('train.csv')
titanic.head()
```

```python
tips = sns.load_dataset('tips')
tips.head(4)
```

```python
sns.get_dataset_names()
```

```python
flights= sns.load_dataset('flights')
flights.head()
```

```python
iris = sns.load_dataset('iris')
iris.head()
```

1. Scatterplot (numerical-numerical)

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue=tips["sex"], style=tips["smoker"])
```

2. Bar plot (categorical-numerical)

```python
sns.barplot(titanic, x= titanic["Pclass"], y= titanic["Age"], hue=titanic["Sex"])

# black line is the confidence interval, which is the range of values that are likely to contain the true population mean with a certain level of confidence (usually 95%). 
# The confidence interval is calculated based on the sample data and provides an estimate of the uncertainty around the mean. 
# If the confidence interval is narrow, it indicates that the sample mean is a more precise estimate of the population mean.
# If the confidence interval is wide, it indicates that there is more uncertainty around the estimate of the population mean.
```

3. Box Plot (categorical-numerical)

```python
sns.boxplot(titanic, x=titanic["Pclass"], y= titanic["Age"], hue=titanic["Survived"])
```

4. Distplot (Numerical-categorical)

```python
sns.distplot(titanic[titanic["Survived"]==0]["Age"], hist=False)
sns.distplot(titanic[titanic["Survived"]==1]["Age"],hist=False)
```
