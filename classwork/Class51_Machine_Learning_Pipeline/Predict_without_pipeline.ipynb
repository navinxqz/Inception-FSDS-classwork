```python
import pickle
import numpy as np
```

```python
ohe_sex=pickle.load(open("models/ohe_sex.pkl","rb"))
ohe_embarked=pickle.load(open("models/ohe_embarked.pkl","rb"))
clf=pickle.load(open("models/clf.pkl","rb"))
```

```python
# Assume user input
# Pclass/gender/age/SibSp/Parch/Fare/Embarked
test_input = np.array([3, 'male', 22, 1, 0, 7.25, 'S'], dtype=object).reshape(1, 7)
```

```python
test_input
```

```python
test_input_sex = ohe_sex.transform(test_input[:, 1].reshape(1, 1))
```

```python
test_input_embarked = ohe_embarked.transform(test_input[:, -1].reshape(1, 1))
```

```python
test_input_embarked
```

```python
test_input_age = test_input[:, 2].reshape(1, 1)
```

```python
test_input_concat = np.concatenate((test_input[:, [0, 3, 4, 5]], test_input_age, test_input_sex, test_input_embarked), axis=1)
```

```python
test_input_concat.shape
```

```python
clf.predict(test_input_concat)

# 1 means survived, 0 means not survived
```
